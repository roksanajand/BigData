Names.csv 
* Dodaj kolumnę z wartością czasu wykonania notatnika w formacie epoch
* Dodaj kolumnę w której wyliczysz wzrost w stopach (feet)
* Odpowiedz na pytanie jakie jest najpopularniesze imię?
* Dodaj kolumnę i policz wiek aktorów 
* Usuń kolumny (bio, death_details)
* Zmień nazwy kolumn - dodaj kapitalizaję i usuń _
* Posortuj dataframe po imieniu rosnąco

```
from pyspark.sql.functions import current_timestamp, unix_timestamp, col,count,split, datediff, when,year,current_date, to_date, round

filePath = "dbfs:/FileStore/tables/Files/names.csv"
namesDf = spark.read.format("csv").option("header","true").option("inferSchema","true").load(filePath)

#display(namesDf)
#Dodaj kolumnę z wartością czasu wykonania notatnika w formacie epoch
df = namesDf.withColumn("current_time", current_timestamp())
df = df.withColumn("epoch_current_time", unix_timestamp(col("current_time")))

#Dodaj kolumnę w której wyliczysz wzrost w stopach (feet)
df = df.withColumn("height_feet",col("height")*0.0328084)

#Odpowiedz na pytanie jakie jest najpopularniesze imię?
df = df.withColumn("first_name", split(col("name"), " ")[0])
first_name_counts = df.groupBy("first_name").count()
most_popular_first_name = first_name_counts.orderBy(col("count").desc()).limit(1)
display(most_popular_first_name)
#John to najpopularniejsze imie

#Dodaj kolumnę i policz wiek aktorów
spark.sql("set spark.sql.legacy.timeParserPolicy=CORRECTED")
df = df.withColumn("bdate",
    when(to_date(col("date_of_birth"), "dd.MM.yyyy").isNotNull(),
         to_date(col("date_of_birth"), "dd.MM.yyyy"))
    .when(to_date(col("date_of_birth"), "yyyy-MM-dd").isNotNull(),
         to_date(col("date_of_birth"), "yyyy-MM-dd"))
    .when(to_date(col("date_of_birth"), "yyyy MM dd").isNotNull(),
         to_date(col("date_of_birth"), "yyyy-MM-dd"))
    .when(to_date(split(col("date_of_birth"), " ").getItem(0), "yyyy").isNotNull(),
         to_date(split(col("date_of_birth"), " ").getItem(0), "yyyy"))
    .otherwise(None)
)
df = df.withColumn("ddate",
    when(to_date(col("date_of_death"), "dd.MM.yyyy").isNotNull(),
         to_date(col("date_of_death"), "dd.MM.yyyy"))
    .when(to_date(col("date_of_death"), "yyyy-MM-dd").isNotNull(),
         to_date(col("date_of_death"), "yyyy-MM-dd"))
    .when(to_date(col("date_of_death"), "yyyy MM dd").isNotNull(),
         to_date(col("date_of_death"), "yyyy-MM-dd"))
    .when(to_date(split(col("date_of_death"), " ").getItem(0), "yyyy").isNotNull(),
         to_date(split(col("date_of_death"), " ").getItem(0), "yyyy"))
    .otherwise(None)
)
df = df.withColumn("current_date", current_date())
df = df.withColumn("age", round(datediff(when(col("ddate").isNotNull(), col("ddate")).otherwise(col("current_date")),col("bdate"))/365.25, 0))
#display(df.select("name", "date_of_birth", "date_of_death","bdate","ddate","current_date", "age"))

#Usuń kolumny (bio, death_details)
df = df.drop("bio", "death_details", "ddate", "bdate", "current_date")

#Zmień nazwy kolumn - dodaj kapitalizaję i usuń _
def format_column_name(name):
    return  name.replace("_", "").capitalize()
for old_name in df.columns:
    df = df.withColumnRenamed(old_name, format_column_name(old_name))

#Posortuj dataframe po imieniu rosnąco
df = df.orderBy(col("name").asc())

df.explain()         
display(df.limit(50))
```

Movies.csv
* Dodaj kolumnę z wartością czasu wykonania notatnika w formacie epoch
* Dodaj kolumnę która wylicza ile lat upłynęło od publikacji filmu
* Dodaj kolumnę która pokaże budżet filmu jako wartość numeryczną, (trzeba usunac znaki walut)
* Usuń wiersze z dataframe gdzie wartości są null

```
from pyspark.sql.functions import current_timestamp, unix_timestamp,regexp_replace
filePath = "dbfs:/FileStore/tables/Files/movies.csv"
namesDf = spark.read.format("csv").option("header","true").option("inferSchema","true").load(filePath)

#Dodaj kolumnę z wartością czasu wykonania notatnika w formacie epoch
df2 = namesDf.withColumn("current_time", current_timestamp())
df2 = df2.withColumn("epoch_current_time", unix_timestamp(col("current_time")))

#Dodaj kolumnę która wylicza ile lat upłynęło od publikacji filmu
spark.sql("set spark.sql.legacy.timeParserPolicy=CORRECTED")

df2 = df2.withColumn("ddate",
    when(to_date(col("date_published"), "dd.MM.yyyy").isNotNull(),
         to_date(col("date_published"), "dd.MM.yyyy"))
    .when(to_date(col("date_published"), "yyyy-MM-dd").isNotNull(),
         to_date(col("date_published"), "yyyy-MM-dd"))
    .when(to_date(col("date_published"), "yyyy MM dd").isNotNull(),
         to_date(col("date_published"), "yyyy-MM-dd"))
    .when(to_date(split(col("date_published"), " ").getItem(0), "yyyy").isNotNull(),
         to_date(split(col("date_published"), " ").getItem(0), "yyyy"))
    .otherwise(None))
df2 = df2.withColumn("current_date", current_date())
df2 = df2.withColumn("age", round(datediff(col("current_date"),col("ddate"))/365.25, 0))
#display(namesDf.select("current_date", "date_published","ddate","current_date", "age"))

#Dodaj kolumnę która pokaże budżet filmu jako wartość numeryczną, (trzeba usunac znaki walut)
df2 = df2.withColumn("budget_numeric", regexp_replace(col("budget"), r"[^\d]", "").cast("double"))
#display(namesDf.select("budget", "budget_numeric"))

#Usuń wiersze z dataframe gdzie wartości są null
df2=df2.dropna()

df2.explain("FORMATTED")
display(df2.limit(50))
```

```
from pyspark.sql.functions import avg
filePath = "dbfs:/FileStore/tables/Files/ratings.csv"
namesDf = spark.read.format("csv").option("header","true").option("inferSchema","true").load(filePath)

#Dodaj kolumnę z wartością czasu wykonania notatnika w formacie epoch
df3 = namesDf.withColumn("current_time", current_timestamp())
df3 = df3.withColumn("epoch_current_time", unix_timestamp(col("current_time")))

df3 = df3.dropna(subset=["males_allages_avg_vote", "females_allages_avg_vote"])
#Kto daje lepsze oceny chłopcy czy dziewczyny dla całego setu
df3 = df3.withColumn("better", when(df3["females_allages_avg_vote"] > df3["males_allages_avg_vote"], "F").when(df3["females_allages_avg_vote"] < df3["males_allages_avg_vote"], "M").otherwise("n/a"))
#df3.groupBy("better").count().show()
#Lepsze oceny dają dziewczyny

#Dla jednej z kolumn zmień typ danych do long
df3 = df3.withColumn("allgenders_45age_votes", df3.total_votes.cast("long"))

#df_grouped = df3.groupBy("better").agg(
  #  avg("males_allages_avg_vote").alias("avg_male_rating"),
   # avg("females_allages_avg_vote").alias("avg_female_rating")
#)
#df_grouped.explain(True)
#display(df_grouped)
df3.explain()
display(df3.limit(50))
```
