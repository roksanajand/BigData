##Zadanie 1

```
from pyspark.sql.functions import (
    col, explode, regexp_replace, regexp_extract,
    coalesce, lit, when, array_contains, array, expr, avg, max, countDistinct
)
```

```
filePath = "dbfs:/FileStore/tables/Files/names.csv"
df = spark.read.format("csv").option("header","true").option("inferSchema","true").load(filePath)

display(df.limit(20))  

df.printSchema()
```

```
df_null_bio = df.filter(col("height").isNull())
display(df_null_bio.limit(20))
```

```
df_filled = df.na.fill({"height": 999})
display(df_filled.limit(20))
                       
```

```
from pyspark.sql import functions as F
df_explode = df_filled.select(F.explode(F.split("Height", '1')))
display(df_explode.limit(20))
                              
```

```
df_drop = df_filled.drop("date_of_death")
display(df_drop.limit(10))
```

```
df_regexp_replace = df_filled.withColumn("new_death_details",regexp_replace("death_details",",",";"))
display(df_regexp_replace.limit(10))
```

```
df_extract = df_filled.withColumn("surname",regexp_extract(col("name"),r"^\S+\s+(\S+)", 1))
df_extract.select("name", "surname").show(10, truncate=False)
```

```
df_ifnull = df_extract.withColumn("place_of_death_ifnull",expr("ifnull(place_of_death, 'UNKNOWN')"))
display(df_ifnull.limit(20))
```

```
df_nullIf = df_ifnull.withColumn("surname_nullIf",expr("nullif(surname, 'Bacall')"))
display(df_nullIf.limit(20))
```

```
df_replace = df_nullIf.replace( ["natural causes"],["NATURAL CAUSES F.E AGE"],"reason_of_death"  )
display(df_replace.limit(10))
```

```

df_array_contains = df_replace.withColumn("has_177",(col("height") == 177))

display(df_array_contains.select("height", "has_177").limit(10))
```

```
df_agg = df_replace.agg(
    avg("spouses").alias("avg_spouses"),
    max("children").alias("max_children"),
    countDistinct("surname").alias("unique_surname")
)

display(df_agg)
```

##Zadanie 2

```
from pyspark.sql.functions import udf
from pyspark.sql.types import StringType


def height_category(h):
    if h is None:
        return "unknown"
    return "tall" if h >= 180 else "short"


height_category_udf = udf(height_category, StringType())

df_with_height_cat = df_replace.withColumn("height_category", height_category_udf(col("height")))

display(df_with_height_cat.select("name", "height", "height_category").limit(10))
```

```
from pyspark.sql.functions import pandas_udf
from pyspark.sql.types import IntegerType
import pandas as pd

@pandas_udf(IntegerType())
def count_words_in_bio(bio_series: pd.Series) -> pd.Series:
    return bio_series.fillna("").apply(lambda x: len(x.split()))

df_with_bio_words = df_replace.withColumn("bio_word_count", count_words_in_bio(col("bio")))

display(df_with_bio_words.select("name", "bio", "bio_word_count").limit(10))
```
